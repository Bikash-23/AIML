```python
import numpy as np 
import pandas as pd
```

<span style="color: orange;">**_Finding Missing Data_**</span>

```python

data = {
    'A': [1, 2, np.nan, 4, 5],
    'B': [1, 2, 3, 4, 5],
    'C': [1, 2, 3, np.nan, np.nan],
    'D': [1, np.nan, np.nan, np.nan, 5]
}
df = pd.DataFrame(data)
```

```python
df
```

```python
df.isna()
```

```python
df.isna().sum()
```

```python
df.isna().any()
```

<span style = "color:orange">**_Removing missing Data_**</span>

```python
df.dropna()
```

```python
# Row wise
# thresh = non null value
df.dropna(thresh=3)
df.dropna(thresh=4)
```

<span style = "color:orange">**_Filling the missing Data_**</span>

```python
df
```

```python
df.fillna(-1)
```

```python
# column wise
val  = {'A':0,'B':-1,'C':-2,'D':-3}
df.fillna(value=val)
```

```python
df.mean()
```

```python
df.fillna(df.mean())
```
