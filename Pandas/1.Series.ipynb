## Pandas Series
A serise is a one-dimentional lebbeled array capable of holding any data type.The axis labels are collectively called the index.

 ## _Creating Series_ 
There are many ways to create a pandas series:

```python
import numpy as np
import pandas as pd
```

```python
labels = ['a','b','c']
lists = [10,20,30]
arr = np.array([10,20,30])
d = {'x':10,'y':20,'z':30}
```

```python
pd.Series(lists)
```

```python
pd.Series(lists,index=labels)
```

```python
pd.Series(arr)
```

```python
pd.Series(d)
```
