NUMPY INDEXING AND SLICING OF VECTORS

```python
arr = np.arange(11,21)
arr
```

```python
arr[6]
```

```python
arr[1:4]
```

```python
arr[:6]
```

```python
arr[2:]
```

```python
arr[1::2]
```

NUMPY INDEXING AND SLICING OF MATRIX

```python
arr = np.arange(1,31).reshape(6,5)
arr
```

```python
#slicing out row
arr[0]
```

```python
#scicing out index
arr[0,0]
```

```python
#slicing out matrix portion
arr[0:2,1:3]
```

```python
arr[3:,3:]
```

```python
#slicing out column
arr[:,2]
```

```python
arr= np.arange(11,21)
arr
```
