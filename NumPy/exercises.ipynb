**Valid Sudoku**

```python
s = np.array([
    [5, 3, 4, 6, 7, 8, 9, 1, 2],
    [6, 7, 2, 1, 9, 5, 3, 4, 8],
    [1, 9, 8, 3, 4, 2, 5, 6, 7],
    
    [8, 5, 9, 7, 6, 1, 4, 2, 3],
    [4, 2, 6, 8, 5, 3, 7, 9, 1],
    [7, 1, 3, 9, 2, 4, 8, 5, 6],
    
    [9, 6, 1, 5, 3, 7, 2, 8, 4],
    [2, 8, 7, 4, 1, 9, 6, 3, 5],
    [3, 4, 5, 2, 8, 6, 1, 7, 9]
])
```

```python
r = np.sum(s,axis = 1)
r == 45
```

```python
c = np.sum(s,axis = 0)
c ==45
```

```python
b = np.sum(s,axis = 1)
for i in b:
    if i != 45:
        print("Suduku is not valid for row")
        break
else:
    print("for rows valid suduku!")
```

```python
c = np.sum(s,axis = 1)
for i in c:
    if i != 45:
        print("Suduku is not valid for columns")
        break
else:
    print("for columns valid suduku!")
```

**General Questions**

```python
import numpy as np

# Columns: [Age, Math Marks, Science Marks]
data = np.array([
    [18, 85, 78],   # Student 1
    [19, 92, 88],   # Student 2
    [17, 76, 95],   # Student 3
    [18, 65, 70],   # Student 4
    [20, 90, 85]    # Student 5
])
```

```python
#Get the shape of the matrix.
data.shape
```

```python
#Find the average age of students.
avg = data[:,0]
# avg.sum()/5
np.mean(avg)
```

```python
#Extract Math marks of all students.
mmark = data[:,1]
mmark
```

```python
#Find the highest Science mark.
smark = data[:,2]
np.max(smark)
```

```python
#Get details of the student who scored more than 90 in Math.
data[mmark > 90]
```

```python
#Increase Math marks of all students by 5.
mmark + 5
```

```python
#Find how many students are younger than 19.
len(data[avg <19])
```

```python
#Calculate the average marks in each subject (column-wise mean).
a = np.mean(mmark)
b = np.mean(smark)
print(a)
print(b)
```

```python
#Get data of students who scored at least 80 in both subjects.
s = data[(mmark >= 80) & (smark >=80)]
s
```

```python
#Replace all Science marks < 75 with 0.
smark[smark<75]  = 0
data
```
