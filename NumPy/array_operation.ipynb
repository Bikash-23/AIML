Deep and Shallow copy

Shallow : main array not change \
Deep: main array change

```python
a = np.arange(1,21)
a
```

```python
#Shallow copy
slice = a[:5]
slice
```

```python
slice*10
```

```python
a
```

```python
a[4:6]*10
```

```python
a
```

```python
#Shallow copy
b = a.copy()
```

```python
b[0] = 99
```

```python
b
```

```python
a
```

```python
#Deep copy
b = a
```

```python
b[0] = 99
```

```python
a
```
