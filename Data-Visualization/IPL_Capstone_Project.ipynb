**Now, Here comes some Basic Questions**

1. Which team won the most matches?

```python
match_winner = df['match_winner'].value_counts()
```

```python
sns.barplot(y = match_winner.index,x = match_winner.values,palette='viridis')
```

2. Toss Decision Trends

```python
df['toss_decision']
```

```python
sns.countplot(x = df['toss_decision'],palette='rainbow')
plt.title('Toss decision trend')
```

3. Toss Winner vs Match Winner

```python
df[df['toss_winner'] == df['match_winner']].head()
```

```python
count = df[df['toss_winner'] == df['match_winner']]['match_id'].count()
percentage = (count*100)/df.shape[0]
percentage.round()
```

4. How do teams win? (Runs vs Wickets)

```python
sns.countplot(x = df['won_by'])
```

###  Key Player Performances

1 Most "Player of the Match" Awards

```python
df['player_of_the_match'].value_counts().head(1)
```

2 Top Scorers

```python
df['highscore'].sort_values(ascending=False).head(2)
```

10 Best Bowling Figures

```python
df['Highest_wickets']=df['best_bowling_figure'].apply(lambda x:x.split('--')[0])
df['Highest_wickets'] = df['Highest_wickets'].astype(int)
df
```

```python
top_bowlers = df.groupby('best_bowling')['Highest_wickets'].sum().sort_values(ascending=False)
top10  = top_bowlers.head(10)
top10
```

```python
top10.plot(kind = 'barh')
```
